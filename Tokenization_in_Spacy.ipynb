```python
doc = nlp('Gimme some extra chess pizza pls tia!')

tokens = [token.text for token in doc]
tokens
```

```python
from spacy.symbols import ORTH

nlp.tokenizer.add_special_case('gimme',[
    {ORTH:'gim'},
    {ORTH:'me'}

])


nlp.tokenizer.add_special_case('tia',[
    {ORTH:'t'},
    {ORTH:'i'},
    {ORTH: 'a'}

])

doc = nlp('gimme some extra chess pizza pls tia!')

tokens = [token.text for token in doc]
tokens
```

```python
doc = nlp('Dr. stanger like pizaa. And Hulk too')

for sentence in doc.sents:
  print(sentence)
```

```python
nlp.add_pipe('sentencizer')

doc = nlp('Dr. stanger like pizaa. And Hulk too')

for sentence in doc.sents:
  print(sentence)
```

(2) Extract all money transaction from below sentence along with currency. Output should be,

two $

500 €



```python
transactions = "Tony gave two $ to Peter, Bruce gave 500 € to Steve"

# TODO: Write code here
transactions = nlp(transactions)
currency = []
for token in transactions:
  if token.is_currency:
    index = token.i - 1
    currency.append(f'{transactions[index]} {token.text}')
# Hint: Use token.i for the index of a token and token.is_currency for currency symbol detection
currency
```

Actual Solution

```python
transactions = "Tony gave two $ to Peter, Bruce gave 500 € to Steve"
doc = nlp(transactions)
for token in doc:
    if token.like_num and doc[token.i+1].is_currency:
        print(token.text, doc[token.i+1].text)
```
