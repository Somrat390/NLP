```python
import spacy

nlp = spacy.load('en_core_web_sm')
```

```python
text = """
Apple announced a new AI product in California.
The company said the product will be available next month.
"""

doc = nlp(text)
```

```python
for token in doc:
  print(token.text)
```

```python
for token in doc:
  print(token.lemma_)
```

```python
for token in doc:
  print(token.text,'->', token.pos_)
```

```python
for token in doc.ents:
  print(token.text, '->', token.label_)
```

```python
import pandas as pd
```

```python
def nlp_analyzer(text):
  doc = nlp(text)

  print("""
  ==========================================
           NewsPaper Analyzer
  ==========================================
  \n\n""")

  print("Article\n", doc)
  print("""\n\n
  ------------------------------------------
            Tokens
  ------------------------------------------
  """)
  for token in doc:
     print(token.text)

  print("""
  ------------------------------------------
            Tokens Analysis
  ------------------------------------------
  """)

  analysis_frame = []
  for token in doc:
    analysis_frame.append({'Token': token.text,
                           'Lemma':token.lemma_,
                           'Pos':token.pos_})

  frame = pd.DataFrame(analysis_frame)
  print(frame)

  print("""\n
  ------------------------------------------
           Name Entities
  ------------------------------------------
  """)


  for token in doc.ents:
    print(token.text, '                   ', token.label_)






```
