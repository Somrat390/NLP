```python
from nltk.stem import PorterStemmer

stemmer = PorterStemmer()
```

```python
words = ['eating', 'eats', 'eat', 'ate', 'adjustabel', 'rafting', 'ability','meeting']

for word in words:
  print(word, ' | ', stemmer.stem(word))
```

```python
import spacy
nlp = spacy.load('en_core_web_sm')

words = nlp('eating eats ate eat adjustable rafting ability meeting better')

for word in words:
  print(word, ' | ', word.lemma_)
```

```python
ar = nlp.get_pipe('attribute_ruler')

ar.add([[{'TEXT':'Bro'}],[{'TEXT':'Brah'}]], {'LEMMA':'Brother'})

doc = nlp("Bro, you wanna go? don't say no! I am exausted")

for token in doc:
  print(token.text, ' | ', token.lemma_)
```
