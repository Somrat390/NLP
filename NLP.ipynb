```python
import re
```

```python
chat1 = 'codebasics: you ask lot of questions 12345678912, abc@xyz.com, 9998881234'
chat2 = 'codebasics: here it is: (123)-567-8912, abc@xyz.com'
chat3 = 'codebasics: yes, phone: 1235678912 email: abc@xyz.com'
```

```python
pattern =r'\d{10}|\(\d{3}\)-\d{3}-\d{4}'
matches = re.findall(pattern, chat2)
matches
```

```python
email_pattern = r'[a-z0-9A-z]+@[a-z0-A-Z]+\.[a-z0-9A-Z]+'
matches = re.findall(email_pattern, chat1)
matches
```

# Spacy

### Download an English language model

```python
!python -m spacy download en_core_web_sm
```

```python
import spacy

nlp = spacy.load("en_core_web_sm")
print("Success!")
```

```python
doc = nlp("Apple is looking at buying a startup in London. I love apple.")

for token in doc:
    print(token.text, token.pos_)
```

```python
for sentence in doc.sents:
  print(sentence)
  for word in sentence:
    print(word)
```
